```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1000,1200,1500,1800,2000], dtype=float)
Y = np.array([50,60,75,90,100], dtype=float)
```

```python
def predict(X,Y,m,b):
  return X*m+b
```

```python
def cost(X,Y,m,b):
  pred = predict(X,Y,m,b)
  error = Y - pred
  return np.mean(error**2)
```

```python
m = 0
h = 0.000000001

for i in range(10000):
  current = cost(X,Y,m, 0)
  newCurrent = cost(X,Y,m+h, 0)

  gradient = (newCurrent - current)/h

  m = m - h*gradient
print(m)
```

```python
learning_rate = 1e-8
epochs = 1000

m = 0
b = 0
```

```python
for epoch in range(epochs):
  predictions = m*X+b

  error = predictions - Y
  cost =np.mean(error**2)

  dm = (2/len(X)) * np.sum(error*X)
  db = (2/len(X)) * np.sum(error)

  m = m - learning_rate*dm
  b = b - learning_rate*db

print(f"Epoch {epoch:4d} | Cost = {cost:.6f} | m = {m:.6f} | b = {b:.6f}")
house_area = 1600
predicted_price = m * house_area + b

print(f"\nPredicted price for {house_area} sq.ft = {predicted_price:.2f} Lakhs")

# Plot Result

plt.scatter(X, Y, color="blue", label="Actual Data")
plt.plot(X, m * X + b, color="red", label="Regression Line")
plt.xlabel("Area")
plt.ylabel("Price")
plt.legend()
plt.show()
```
