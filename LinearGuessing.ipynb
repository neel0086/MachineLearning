Imports


```python
import numpy as np
import matplotlib.pyplot as plt
```

DATASETS

```python
X = np.array([1000, 1200, 1500, 1800, 2000])
Y = np.array([50, 60, 75, 90, 100])
```

Visualize DataSet

```python
plt.scatter(X,Y)
plt.xlabel("Area")
plt.ylabel("Price")
plt.show()
```

Guess Values

```python
m = 0.03
b = 20

predicted = m*X + b

print("Predictions:", predicted,"\n")

for actual, pred in zip(Y, predicted):
  print(f"Actual = {actual}, Predcited = {pred}")

```

```python
error = Y - predicted

squared_error = error**2

squared_error
```

```python
mse = np.mean(squared_error)
mse
```
