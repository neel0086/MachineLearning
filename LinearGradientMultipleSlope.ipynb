```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [1000,2,10,1],
    [1200,3,7,1],
    [1500,3,4,2],
    [1800,4,2,2]
], dtype=float)

Y = np.array([50,60,75,90], dtype=float)
```

```python
weights = np.zeros(X.shape[1])
bias = 0

learning_rate = 5e-7
epochs = 1000
```

```python
for epoch in range(epochs):
  prediction = X@weights + bias
  error = prediction - Y
  cost = np.mean(error**2)
  dw = (2/len(X))*(X.T @ error)
  db = (2/len(X))*np.sum(error)

  weights -= learning_rate*dw
  bias -= learning_rate*db
  if epoch % 100 == 0:
    print("----------------")
    print("Epoch:", epoch)
    print("Weights:", weights)
    print("dw:", dw)
    print("cost:", cost)
```
