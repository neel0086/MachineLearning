```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [2, 22],
    [3, 25],
    [5, 30],
    [7, 35],
    [8, 40],
    [9, 45]
], dtype=float)

Y = np.array([0, 0, 0, 1, 1, 1], dtype=float)

#Scaling lets say first feild is 100000, and second is 10, than answer will be dominated by first feild so scaling them between 0 to 1
X = X/np.max(X, axis=0)
```

```python
split = int(len(X)*0.8)

X_train = X[:split]
Y_train = Y[:split]
X_test = X[split:]
Y_test = Y[split:]
```

```python
weights = np.zeros(X_train.shape[1])

bias = 0
learning_rate = 0.1

epochs = 5000
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
def loss(y_true, y_pred):
  eps = 1e-15
  y_pred = np.clip(y_pred, eps, 1-eps)
  return -np.mean(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))
```

```python
for epoch in range(epochs):
  z = X_train@weights + bias

  prediction = sigmoid(z)

  current_loss = loss(Y_train, prediction)
  error = prediction - Y_train

  dw = (1/len(X))*(X_train.T@error)
  db = (1/len(X))*np.sum(error)

  weights -= learning_rate*dw
  bias -= learning_rate*db

print("\nWeights:", weights)
print("Bias:", bias)


```

```python
#TESTING
z = X_test@weights + bias
probability = sigmoid(z)

prediction = (probability>=0.5).astype(int)

print("\nProbabilities:", probability)
print("Prediction:", prediction)
print("Actual:", Y_test)

accuracy = np.mean(prediction == Y_test)

print("\nAccuracy =", accuracy*100,"%")
```
